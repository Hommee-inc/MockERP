# Provider Contract

## Cell 1 — Imports

```python
from abc import ABC, abstractmethod
```

## Cell 2 — Name Provider

```python
class NameProvider(ABC):

    @abstractmethod
    def name(self):
        pass
```

## Cell 3 — Email Provider

```python
class EmailProvider(ABC):

    @abstractmethod
    def email(self):
        pass
```

## Cell 4 — Address Provider

```python
class AddressProvider(ABC):

    @abstractmethod
    def address(self):
        pass
```

## Cell 5 — Phone Provider

```python
class PhoneProvider(ABC):

    @abstractmethod
    def phone(self):
        pass
```

## Cell 6 — Document Provider

```python
class DocumentProvider(ABC):

    @abstractmethod
    def document(self, document_type):
        pass
```
